```python
%cd /kaggle/working/AffectNet2013

from pathlib import Path
import re

p = Path("predict.py")
s = p.read_text()

# Replace the output path with Kaggle's writable working directory
s, count = re.subn(
    r'^\s*out = Path\(path\).*?$',
    '        out = Path("/kaggle/working") / (Path(path).stem + "_prediction.png")',
    s,
    flags=re.MULTILINE
)

assert count == 1, f"Expected to change one output path, changed {count}"

p.write_text(s)

# Verify the corrected path
for line in s.splitlines():
    if "out = Path(" in line:
        print("Corrected save path:", line.strip())

# Run prediction again
!python predict.py \
    --checkpoint runs_ex4d_100/mek_resnet18_ferplus_majority_224px_sqrt_inverse/best.pt \
    /kaggle/input/datasets/mayank8636/fer-my-facial-expressions/Capture.PNG

# Verify saved image
output = Path("/kaggle/working/Capture_prediction.png")
print("Prediction image saved:", output.exists())
```
